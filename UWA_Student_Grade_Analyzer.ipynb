```python
import pandas as pd

data = {
    "Student": ["Aakarsh", "Bharat", "Ayush", "Atharva", "Akanksha"],
    "Python": [75, 68, 45, 82, 60],
    "RDBMS": [70, 72, 50, 85, 58],
    "Statistics": [65, 74, 55, 78, 62],
    "OpenSource": [80, 70, 60, 88, 66]
}

df = pd.DataFrame(data)
df
```

```python
df["Average"] = df[
    ["Python", "RDBMS", "Statistics", "OpenSource"]
].mean(axis=1)

df
```

```python
def grade(mark):
    if mark < 50:
        return "N"
    elif mark < 60:
        return "P"
    elif mark < 70:
        return "CR"
    elif mark < 80:
        return "D"
    else:
        return "HD"
```

```python
df["Grade"] = df["Average"].apply(grade)
df
```

```python
df["Status"] = df["Average"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)

df
```

```python
top_student = df.loc[df["Average"].idxmax()]
top_student
```

```python
lowest_student = df.loc[df["Average"].idxmin()]
lowest_student
```

```python
subject_avg = df[
    ["Python","RDBMS","Statistics","OpenSource"]
].mean()

subject_avg
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(df["Student"], df["Average"])

plt.title("Student Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")

plt.show()
```

```python
df.to_csv(
    "student_grade_report.csv",
    index=False
)
```
